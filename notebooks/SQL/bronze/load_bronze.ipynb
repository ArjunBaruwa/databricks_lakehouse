# **Load Bronze Layer (Source -> Bronze)**


### **Purpose**: 
	This notebook loads data into the bronze schema from external CSV files.
	This performs below tasks:
		- Truncate bronze tables before loading data
		- Uses 'Insert Insert' to load data from the files to the tables


```python
%sql
TRUNCATE TABLE datalakehouse.bronze.crm_cust_info;
INSERT INTO datalakehouse.bronze.crm_cust_info
SELECT cst_id, cst_key, cst_firstname, cst_lastname, cst_marital_status, cst_gndr, cst_create_date
FROM read_files('/Volumes/datawarehouse/bronze/data/cust_info.csv');
        
TRUNCATE TABLE datalakehouse.bronze.crm_prd_info;
INSERT INTO datalakehouse.bronze.crm_prd_info
SELECT prd_id, prd_key, prd_nm AS prd_name, prd_cost, prd_line, prd_start_dt, prd_end_dt
FROM read_files('/Volumes/datawarehouse/bronze/data/prd_info.csv');
        
TRUNCATE TABLE datalakehouse.bronze.crm_sales_details;
INSERT INTO datalakehouse.bronze.crm_sales_details
SELECT sls_ord_num, sls_prd_key, sls_cust_id, sls_order_dt, sls_ship_dt, sls_due_dt, sls_sales, sls_quantity, sls_price
FROM read_files('/Volumes/datawarehouse/bronze/data/sales_details.csv');
        
```
