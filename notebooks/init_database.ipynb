# **Create Catalog and Schemas**

### # Script Purpose:
    This script creates a new catalog named 'DataLakehouse' after checking if it already exists. 
    If the catalog exists, it is dropped and recreated. Additionally, the script sets up three schemas 
    within the catalog: 'bronze', 'silver', and 'gold'.
	
### # WARNING:
    Running this script will drop the entire 'DataLakehouse' catalog if it exists. 
    All data in the catalog will be permanently deleted. Proceed with caution 
    and ensure you have proper backups before running this script.


```python
%sql
--Below line drops the catalog if exists
DROP CATALOG IF EXISTS datalakehouse CASCADE;
CREATE CATALOG IF NOT EXISTS datalakehouse;
```

```python
%sql
--Below line drops bronze schema if exists
DROP SCHEMA IF EXISTS datalakehouse.bronze CASCADE;
CREATE SCHEMA IF NOT EXISTS datalakehouse.bronze;

--Below line drops silver schema if exists
DROP SCHEMA IF EXISTS datalakehouse.silver CASCADE;
CREATE SCHEMA IF NOT EXISTS datalakehouse.silver;

--Below line drops gold schema if exists
DROP SCHEMA IF EXISTS datalakehouse.gold CASCADE;
CREATE SCHEMA IF NOT EXISTS datalakehouse.gold;
```
